# Metaheurystyki

Heurystyka to praktyczne podejście do rozwiązywania problemów obliczeniowych, które stawia na szybkość i efektywność kosztem gwarancji znalezienia optymalnego wyniku. W przypadku trudnych problemów kombinatorycznych lub NP-trudnych, wyznaczenie ścisłego ekstremum funkcji celu metodami przeglądu wyczerpującego czy algorytmami gradientowymi bywa niemożliwe z powodu ogromnej złożoności czasowej. Heurystyka działa w oparciu o odgórnie ustalone reguły, które pozwalają przeszukiwać przestrzeń rozwiązań w sposób ukierunkowany. Jej głównym celem jest dostarczenie w akceptowalnym czasie wyniku „wystarczająco dobrego”, czyli bliskiego optimum.

Szczególną zaletą metod heurystycznych jest ich wysoka elastyczność oraz odporność na specyficzne ograniczenia matematyczne, takie jak brak różniczkowalności czy ciągłości funkcji celu. Podczas gdy klasyczne algorytmy dedykowane konkretnym problemom wykorzystują specyficzną wiedzę o ich strukturze, metaheurystyki (np. algorytmy ewolucyjne) stanowią uniwersalne szkielety obliczeniowe. Łączą one w sobie mechanizmy eksploracji (przeszukiwania nowych obszarów przestrzeni) i eksploatacji (ulepszania już znalezionych obiecujących stanów), co zapobiega przedwczesnemu utknięciu w minimach lokalnych. Ceną za tę wszechstronność jest brak precyzyjnego oszacowania błędu oraz konieczność starannego doboru parametrów sterujących pracą algorytmu.

## Algorytmy ewolucyjne

Algorytmy ewolucyjne (ang. Evolutionary Algorithms) to szeroka rodzina uniwersalnych metaheurystyk przeszukiwania i optymalizacji, których działanie opiera się na podstawowych mechanizmach biologicznej ewolucji. Sercem tych metod jest Darwinowska koncepcja doboru naturalnego, w której słabsze jednostki ulegają eliminacji, a osobniki lepiej przystosowane do warunków środowiska mają wyższe prawdopodobieństwo przetrwania i przekazania swoich cech kolejnym pokoleniom. Proces ten napędzany jest przez dziedziczenie oraz stałe wprowadzanie nowej zmienności genetycznej poprzez mutacje i wymianę materiału genetycznego.

## Algorytmy genetyczne

Algorytmy genetyczne (ang. Genetic Algorithms) stanowią najpopularniejszą podklasę algorytmów ewolucyjnych. Ich cechą wyróżniającą jest ścisłe rozgraniczenie pomiędzy genotypem (kodowaną reprezentacją rozwiązania) a fenotypem (konkretnym rozwiązaniem w przestrzeni problemu). Podczas gdy ogólne algorytmy ewolucyjne mogą operować bezpośrednio na strukturach danych dziedziny problemu, algorytmy genetyczne kładą nacisk na jawne operatory krzyżowania oraz mutacji wykonywane na poziomie zakodowanego genomu, wykorzystując funkcję przystosowania jako odpowiednik środowiskowej presji selekcyjnej.

> Paradygmat obliczeń inspirowanych naturą wykracza jednak poza samą ewolucję biologiczną i genetyczną, obejmując również ewolucyjne metody inteligencji rojowej (ang. Swarm Intelligence). Metody te czerpią inspirację ze zbiorowego, samoorganizującego się zachowania populacji organizmów, takich jak ławice ryb, stadka ptaków czy owady społeczne, z czego można wyróżnić następujące podejścia:
> - Algorytmy Roju Cząstek (ang. Particle Swarm Optimization - PSO),
> - Algorytmy Pszczele (np. Artificial Bee Colony - ABC),
> - Algorytmy Mrówcze (ang. Ant Colony Optimization - ACO).
>
> Chociaż algorytmy rojowe nie stosują klasycznych operatorów genetycznych (krzyżowania czy mutacji), dzielą z algorytmami ewolucyjnymi tę samą podstawową filozofię: operują na populacji potencjalnych rozwiązań i wykorzystują proste, lokalne reguły do wyłaniania złożonego, optymalnego zachowania globalnego.

### Podstawowe pojęcia

**Osobnik** (ang. individual) - pojedynczy reprezentant populacji stanowiący jedno konkretne, potencjalne rozwiązanie rozważanego problemu optymalizacyjnego.

**Populacja** (ang. population) - zbiór wszystkich osobników istniejących w danej generacji (iteracji) algorytmu. Populacja ewoluuje jako całość, podlegając procesom selekcji, krzyżowania i mutacji.

**Genotyp** - wewnętrzna, zakodowana reprezentacja osobnika (struktura danych, na której bezpośrednio operują operatory genetyczne, np. ciąg bitów, wektor liczb rzeczywistych lub ciąg permutacji).

**Fenotyp**: - dekodowana, jawna postać osobnika w przestrzeni problemu. Czyli konkretne rozwiązanie gotowe do oceny (np. konkretny układ trasy w problemie komiwojażera).

**Gen** - najmniejsza jednostka informacyjna w genomie, kodująca pojedynczą cechę lub zmienną decyzyjną problemu.

**Funkcja Przystosowania** (ang. fitness function) - matematyczna miara jakości osobnika, która przypisuje każdemu rozwiązaniu wartość liczbową określającą, jak dobrze radzi sobie ono w danym środowisku. Odpowiada funkcji celu w tradycyjnej optymalizacji i stanowi podstawę dla presji selekcyjnej.

**Generacja** (iteracja, pokolenie) - pojedynczy krok pętli głównej algorytmu ewolucyjnego, w którym dotychczasowa populacja przekształca się w nową populację w wyniku działania operatorów selekcji i ewolucyjnych.

### Selekcja

Operator selekcji (ang. selection) odpowiada za stworzenie presji ewolucyjnej. Jej zadaniem jest wybór osobników, które staną się rodzicami dla nowej generacji, przy czym osobniki o wyższym przystosowaniu mają większe szanse na reprodukcję.

#### Selekcja losowa 

Metoda selekcji losowej (ang. random selection) to najprostsza metoda wyboru osobników i polega na przydzieleniu każdemu osobnikowi dokładnie takiej samej szansy na wybór, niezależnie od wartości funkcji przystosowania. Metoda ta nie wywiera żadnej presji selekcyjnej i jest stosowana głównie jako punkt odniesienia w badaniach lub jako element wspomagający w celu utrzymania różnorodności populacji.

1. Określenie $N$ osobników, które mają zostać wybrane do grupy rodzicielskiej lub nowej populacji.
2. Dla każdego z $N$ miejsc w nowej populacji losowany (ze zwracaniem) jest pojedynczy osobnik z aktualnej populacji bazowej przy użyciu rozkładu jednostajnego.
3. Prawdopodobieństwo wyboru dowolnego $i$-tego osobnika jest równe i wynosi $P(i) = \frac{1}{M}$, gdzie $M$ to rozmiar populacji bazowej.
4. Proces kończy się w momencie skompletowania wymaganej liczby $N$ osobników.

#### Selekcja ruletkowa

Metoda selekcji ruletkowej (ang. roulette wheel selection) polega na przydzieleniu prawdopodobieństwa wyboru osobnika w postaci wprost proporcjonalnej do jego wartości funkcji przystosowania $f(x)$. Rozwiązanie takie prowadzi jedna do wczesnej dominacji "super osobnika", który szybko przejmuje całą populację i prowadzi do przedwczesnej zbieżności, oraz spadek presji selekcyjnej w późniejszych etapach, gdy wartości przystosowania osobników stają się wyrównane. W praktyce prowadzi to najczęściej do utknięcia w minimach lokalnych.

W praktyce metoda może być interpretowana jako koło ruletki, gdzie wielkości pól są proporcjonalne do wartości funkcji przystosowania danego osobnika.

1. Dla każdego z $M$ osobników w populacji obliczana jest wartość funkcji przystosowania $f_i: f_i > 0$.
2. Dla każdego osobnika i obliczane jest jego prawdopodobieństwo wyboru $p_i$ stanowiące udział jego przystosowania w sumie całkowitej: $p_i = \frac{f_i}{S}$, gdzie $S = \sum_{i=1}^{M} f_i$.
3. Wyznaczenie skumulowanego prawdopodobieństwa (dystrybuanty) $q_i$ dla każdego osobnika, co pozwala podzielić odcinek $[0,1]$ na $M$ ciągłych, rozłącznych przedziałów: $q_i = \sum_{j=1}^i p_j: 1 \leq i \leq M$.
4. Ustawienie licznika wybranych osobników i określenie docelowy rozmiar grupy rodzicielskiej $N$.
5. Wykonanie $N$ razy, ze zwracaniem, aż do skompletowania całej nowej populacji rodzicielskiej.
   1. Losowanie w każdej iteracji wartości pseudolosowej (wskazówka ruletki) $r$ z rozkładu jednostajnego w przedziale $[0,1)$.
   2. Wybór osobnika $i$, dla którego wskaźnik $r$ wpada w jego przedział, czyli spełniony jest warunek: $q_{i-1} \leq r \leq q_i: q_0 = 0$.

#### Stochastyczne uniwersalne próbkowanie

Metoda stochastycznego uniwersalnego próbkowania (ang. Stochastic Universal Sampling - SUS) to deterministyczno-losowa modyfikacja klasycznej selekcji ruletkowej. Głównym celem metody SUS jest wyeliminowanie błędu próbkowania , czyli sytuacji, w której rzeczywiście wybrana liczba kopii danego osobnika znacząco odbiega od jego wartości oczekiwanej wynikającej z funkcji przystosowania.

W praktyce metoda może być interpretowana jako koło ruletki z $N$ równomiernie rozłożonymi wskazówkami.

1. Wyznaczenie dystrybuanty $q_i$ dla każdego osobnika, co pozwala podzielić odcinek $[0,1]$ na $M$ ciągłych, rozłącznych przedziałów: $q_i = \sum_{j=1}^i p_j: 1 \leq i \leq M$.
2. Wyznaczenie odległości między wskazówkami (koła ruletki) $\sigma = \frac{1}{N}$, gdzie $N$ to liczba osobników do wyboru.
3. Wyznaczanie wartości startowej $r$ z rozkładu jednostajnego w przedziale $[0;r)$.
4. Wyznaczanie pozycji wskazówek w postaci wektora $N$-elementowego: $p_k = r + (k-1) \cdot \sigma: 1 \leq k \leq N$.
5. Selekcja osobników poprzez umieszczenia $p_k$ w przedziałach poszczególnych osobników.

#### Selekcja rangowa

Metoda selekcji rangowej (ang. rank selection) została zaprojektowana w celu wyeliminowania wad selekcji ruletkowej, czyli przedwczesnej zbieżności (związanego z dominacją „super osobników”) oraz spadku presji selekcyjnej pod koniec działania algorytmu, gdy wartości funkcji przystosowania osobników ulegają wyrównaniu.

W selekcji rangowej bezpośrednia wartość funkcji przystosowania $f(x)$ jest ignorowana. Osobnik otrzymuje natomiast rangę $r \in \{1, 2, ..., N\}$ na podstawie swojej pozycji w posortowanej populacji, gdzie $r=1$ to najgorszy osobnik, a $r=N$ - najlepszy. Prawdopodobieństwo wyboru osobnika zależy więc wyłącznie od przypisanej rangi, co gwarantuje stałą i kontrolowaną presję selekcyjną niezależnie od bezwzględnych wartości funkcji celu.

Metoda selekcji rangowej wsytępuje w dwóch wariantach: liniowym oraz wykładniczym. W wariancie liniowym, prawdopodobieństwo rośnie liniowo wraz z rangą, co pozwala płynnie kontrolować presję selekcyjną za pomocą jednego parametru. W wariancie wykładniczym prawdopodobieństwo rośnie wykładniczo wraz z rangą, co silniej faworyzuje ścisłą czołówkę populacji, drastycznie ograniczając szanse słabszych osobników.

1. Sortowanie populacji $P$ według wartości funkcji przystosowania.
2. Przypisanie każdemu osobnikowi rangi $r \in \{1, 2, ..., N\}$.
3. Wyznaczenie prawdopodobieństwa wyboru dla każdej rangi
   1. Liniowy: $p(r) = \frac{1}{N} \left( (2-s) + 2(s-1) \frac{r-1}{N-1} \right)$, gdzie $s \in [1, 2)$ - oczekiwana liczba kopii najlepszego osobnika.
   2. Wykładniczy: $p(r) = \frac{c^{N-r}}{\sum_{j=1}^N c^{N-j}} = \frac{C^{N-r} \times (1-c)}{1-c^N}$, gdzie $c \in (0,1)$ - skala wykładnicza.
4. Wyznaczenie dystrybuanty dla każdego osobnika od $i=1$ do $N$: $q_i = \sum_{j=1}^i p(j): q_0 = 0, q_n = 1$.
5. Utworzenie pustej populacji rodzicielskiej $P_{parents}$.
6. Powtórzenie $N$ razy:
   1. wylosowanie wartości $u \sim U[0, 1)$,
   2. znalezienie indeksu $k: q_{k-1} \leq u < q_k$,
   3. dodanie osobnika o randze $k$ do $P_{parents}$.

#### Selekcja turniejowa

Metoda selekcji turniejowej (ang. tournament selection) należy do najpopularniejszych metod selekcji stosowanych w algorytmach genetycznych. Łączy w sobie prostotę implementacji, niską złożoność obliczeniową oraz możliwość łatwego kontrolowania presji selekcyjnej. Selekcja turniejowa zamiast analizować całą populację naraz (jak w selekcji ruletkowej czy rangowej), organizuje małe, lokalne „turnieje”. Z aktualnej populacji losuje się $k$ osobników, z których wygrywa jeden o najlepszym przystosowaniu i przechodzi do grupy rodzicielskiej. Proces ten jest powtarzany w pętli aż do skompletowania całej nowej populacji.

1. Ustalenie rozmiaru turnieju $k$ - zazwyczaj 2-5 osobników.
2. Pętla selekcji powtarzania $N$ razy:
   1. losowanie $k$ uczstników z rozkładu jednostajnego,
   2. ocena poprzez porównanie wartości funkcji przystosowania $f(x)$ dla wylosowanych $k$ osobników,
   3. wyłonienie zwycięzcy, dla którego wartość $f(x)$ jest największa,
   4. dodanie zwycięzcy do nowej populacji rodzicielskiej, gdzie oryginalny osobnik pozostaje nadal w populacji bazowej i może brać udział w kolejnych turniejach.

Rozmiar turnieju k bezpośrednio decyduje o sile presji ewolucyjnej, gdzie:
- $k=1$ to odpowiednik selekcji całkowicie losowej przy całkowitym braku presji ewolucyjnej,
- $k=2$ to najczęściej stosowany wariant, który zapewnia umiarkowaną presję selekcyjną, dając słabszym osobnikom realną szansę na przetrwanie, o ile jeśli w turnieju trafią na jeszcze słabszego rywala.
- Duże $k$, np. $k \geq 5$ to silna presja selekcyjna, gdzie szanse słabszych osobników drastycznie spadają, co przyspiesza zbieżność algorytmu, ale zwiększa ryzyko utknięcia w optimum lokalnym.

### Krzyżowanie

Operator krzyżowania (ang. crossover) łączy materiał genetyczny dwóch (lub więcej) osobników rodzicielskich, tworząc nowe rozwiązania (potomków), które dziedziczą cechy obojga rodziców. Krzyżowanie w algorytmach genetycznych operuje w warstwie eksploracji przestrzeni poszukiwań. Ich podstawowym zadaniem jest wymiana materiału genetycznego (genów) pomiędzy osobnikami rodzicielskimi w celu stworzenia potomków, którzy dziedziczą unikalne kombinacje cech obojga rodziców.

Sposób realizacji krzyżowania zależy bezpośrednio od wybranej reprezentacji genomu.

#### Operatory dla reprezentacji binarnej i dyskretnej

1. **Krzyżowanie Jednopunktowe**  (ang. single-point crossover)
- losowany jest jeden punkt cięcia, jako pozycja w genomie,
- potomek otrzymuje pierwszy segment genomu od pierwszego rodzica, a drugi segment od drugiego rodzica (drugi potomek powstaje z odwrotnej kombinacji).

image

2. **Krzyżowanie wielopunktowe** (ang. multi-point crossover)
- losowane są minimum dwa punkty cięcia,
- genom dzielony jest na naprzemienne odcinki, które są wymieniane między rodzicami. Chroni to przed rozbijaniem zebranych cech znajdujących się na skrajnych końcach genomu.

image

3. **Krzyżowanie jednorodne** (ang. uniform crossover)

Nie wyznacza się punktów cięcia. Dla każdego genu indywidualnie losuje się (zazwyczaj z prawdopodobieństwem 50%), od którego z rodziców zostanie on przekazany do potomka.

#### Operatory dla reprezentacji rzeczywistej

W optymalizacji ciągłej, gdzie genomem jest wektor liczb $x \in R^n$, zamiana odcinków genomu byłaby zbyt drastyczna. W taki prszypadkach stosuje się więc operatory uśredniające i kombinatoryczne.

1. **Krzyżowanie arytmetyczne / uśredniające**  (ang. arithmetic crossover)
Potomek powstaje jako kombinacja liniowa wektorów rodziców $R_1$ i $R_2$:
- $\text{Potomek}_1 = \alpha \times R_1 + (1 - \alpha) \times R_2$
- $\text{Potomek}_2 = (1 - \alpha) \times R_1 + \alpha \times R_2$

gdzie $\alpha \in [0, 1]$ - może być stałą wartością lub liczbą losowaną dla każdego genu.

2. **Krzyżowanie Blend** (ang. blend crossover - BLX-α)

Umożliwia generowanie wartości spoza przedziału wyznaczonego bezpośrednio przez rodziców. Nowa wartość genu $x$ jest losowana z rozszerzonego przedziału $[ x_{min} - I \times \alpha, x_{max} + I \times \alpha ]$, gdzie:
- $I = [ x_1 - x_2 ]$,
- $\alpha$ - rozszerzenie, zwykle $\alpha = 0.5$.

### Mutacja 

Operator mutacji (ang. mutation) wprowadza losowe, zazwyczaj drobne zmiany w genomie pojedynczego osobnika z pewnym (zwykle niskim) prawdopodobieństwem. Mutacja w algorytmach genetycznych operuje w warstwie eksploatacji przestrzeni poszukiwań. Operator mutacji odpowiada za wprowadzanie nowej zmienności genetycznej do populacji. O ile krzyżowanie skupia się na eksploracji poprzez kombinowanie istniejących już cech rodziców, o tyle mutacja odpowiada za lokalną eksploatację / dywersyfikację. Zapobiega to utracie materiału genetycznego i chroni algorytm przed przedwczesnym utknięciem w lokalnym optimum.

Sposób realizacji mutacji zależy bezpośrednio od wybranej reprezentacji genomu.

#### Operatory dla reprezentacji binarnej

1. **Negacja bitu** (ang. bit-flip mutation)

Dla każdego genu (bitu) w genomie losowana jest wartość z rozkładu jednostajnego. Jeśli wartość ta jest mniejsza od $p_m$, stan bitu jest odwracany: $1 \rightarrow 0$ lub $0 \rightarrow 1$.

2. **Mutacja punktowa** (ang. single-point mutation)

Zamiast sprawdzać każdy bit po kolei, losuje się dokładnie jedną pozycję w genomie (locus) i odwraca wartość znajdującego się tam bitu.

#### Operatory dla reprezentacji rzeczywistej

Gdy genom jest wektorem liczb rzeczywistych $x \in R^n$ operowanie wyłącznie prostym zamienianiem wartości byłoby zbyt drastyczne lub nieefektywne. Stosuje się zatem modyfikacje arytmetyczne.

1. **Mutacja Gaussowska** (ang. Gaussian mutation)

Do wartości wybranego genu dodawana jest zmienna losowa o rozkładzie normalnym $N(0, \sigma)$: $x_i^{"} = x_i + N(0, \sigma)$, gdzie wartość odchylenia standardowego $\sigma$ decyduje o zasięgu mutacji.

2. **Mutacja równomierna** (ang. uniform mutation)

Wartość modyfikowanego genu jest zastępowana nową liczbą losowaną z rozkładu jednostajnego w dopuszczalnym przedziale $[x_{min}, x_{max}]$.

3. **Mutacja nierównomierna** (ang. non-uniform / creep mutation)

Modyfikacja genu o małą wartość losową, której promień maleje wraz ze wzrostem liczby generacji, co sprzyja precyzyjnej lokalnej optymalizacji na końcowych etapach pracy algorytmu.

Definicja funkcji celu, poddawanej maksymalizacji

```python
def objective_function(x: float, y: float) -> float:
    return np.sin(x) * np.cos(y) + np.exp(-(x**2 + y**2) / 10.0)
```

Definicja funkcji przystosowania, jako adapter dla zdefiniowanej funkcji celu.

```python
def fitness_func(ga_instance: Any, solution: Iterable, solution_idx: Any) -> float:
    x, y = solution[0], solution[1]
    return objective_function(x, y)
```

Pomocniczy callback przechwytujący i zapisujący pozycje populacji w każdej generacji algorytmu

```python
population_history = []
```

```python
def on_generation(ga_instance: Any) -> None:
    population_history.append(ga_instance.population.copy())
```

Konfiguracja algorytmu genetycznego

```python
ga_instance = pygad.GA(
    num_generations=50,             # liczba pokoleń (generacji)
    num_parents_mating=10,          # liczba rodziców wybranych do krzyżowania
    fitness_func=fitness_func,      # zdefiniowana funkcja celu
    sol_per_pop=40,                 # 40 osobników w populacji
    num_genes=2,                    # dwie zmienne: x oraz y - długość genomu
    init_range_low=-5.0,            # zakres początkowy dla wartości genów
    init_range_high=5.0,            # zakres końcowy dla wartości genów
    random_mutation_min_val=-0.5,   # początek zakresu zmienności mutacji
    random_mutation_max_val=0.5,    # koniec zakresu zmienności mutacji
    parent_selection_type="tournament",  # selekcja turniejowa
    K_tournament=3,                 # rozmiar turnieju
    crossover_type="single_point",  # krzyżowanie jednopunktowe
    mutation_type="random",         # mutacja losowa
    mutation_percent_genes=50,      # szansa mutacji dla genu (%)
    gene_space=[                    # zakres zmienności przestrzeni wartości genów (analogicznie do: random_mutation_min_val / max)
        {"low": -5.0, "high": 5.0},
        {"low": -5.0, "high": 5.0},
    ],
    on_generation=on_generation,    # callback (funkcja) wykonująca się "pod spodem" podczas kazdej generacji
)
```

Zachowanie pozycji populacji początkowej

```python
population_history.append(ga_instance.initial_population.copy())
```

Uruchomienie ewolucji

```python
ga_instance.run()
```

Uzyskanie najlepszego rozwiązania

```python
solution, solution_fitness, _ = ga_instance.best_solution()
```

```python
print(f"Znalezione maksimum w punkcie (x, y): ({solution[0]:.4f}, {solution[1]:.4f})")
print(f"Wartość funkcji w tym punkcie: {solution_fitness:.4f}")
```

Wizualiacja przebiegu optymalizacji

```python
fig, ax = plt.subplots(figsize=(8, 6))

x_grid = np.linspace(-5, 5, 200)
y_grid = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x_grid, y_grid)
Z = objective_function(X, Y)

contour = ax.contourf(X, Y, Z, levels=30, cmap="viridis")
fig.colorbar(contour, label="Wartość $f(x, y)$")

scat = ax.scatter([], [], c="red", edgecolors="white", s=50, label="Osobniki (Geny)")
title = ax.set_title("")

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_xlabel("Gen X")
ax.set_ylabel("Gen Y")
ax.legend(loc="upper left")


# Funkcja inicjalizująca animację
def init() -> tuple:
    scat.set_offsets(np.empty((0, 2)))
    return scat, title

# Funkcja aktualizująca klatkę animacji
def update(frame):
    pop = population_history[frame]
    scat.set_offsets(pop[:, :2])
    title.set_text(f"Generacja {frame}/{len(population_history)-1}")
    return scat, title


ani = FuncAnimation(
    fig, update, frames=len(population_history),
    init_func=init, interval=150, blit=True, repeat=True
)

plt.tight_layout()
plt.close()
HTML(ani.to_jshtml())
```

### Przykład 2: minimalizacja funkcji sześcioparametrowej

Spróbujmy znaleźć optymalny zestaw sześciu parametrów ($x_1 - x_6$), który zminimalizuje różnicę względem równania:
$11 x_1 + 2.2 x_2 + (-0.7) x_3 + 4 x_4 + 8.9 x_5 + (-1) x_6 = 44$.

Dane wejściowe i wyjściowe optymalizowanej funkcji 

```python
function_inputs = [11, 2.2, -0.7, 4, 8.9, -1]
desired_output = 44
```

Definicja funkcji przystosowania

```python
def fitness_func(ga_instance: Any, solution: Iterable, solution_idx: Any) -> float:
    # wynik dla danego osobnika
    output = np.sum(solution * function_inputs)
    # wartość funkcji fitness jako abs
    fitness = 1.0 / (np.abs(output - desired_output) + 0.000001)
    return fitness
```

Konfiguracja heurystyki  - algorytmu genetycznego

```python
ga_instance = pygad.GA(
    num_generations=100,             # liczba pokoleń (generacji)
    num_parents_mating=4,            # liczba rodziców wybranych do krzyżowania
    fitness_func=fitness_func,       # zdefiniowana funkcja celu
    sol_per_pop=10,                  # rozmiar populacji (liczba osobników)
    num_genes=len(function_inputs),  # długość genomu (liczba zmiennych)
    init_range_low=-10,              # zakres początkowy dla wartości genów
    init_range_high=10,              # zakres końcowy dla wartości genów
    parent_selection_type="tournament", # selekcja turniejowa
    K_tournament=3,                  # rozmiar turnieju
    crossover_type="single_point",   # krzyżowanie jednopunktowe
    mutation_type="random",          # mutacja losowa
    mutation_percent_genes=20,       # szansa mutacji dla genu (%)
)
```

Uruchomienie ewolucji

```python
ga_instance.run()
```

Wyniki

```python
solution, solution_fitness, solution_idx = ga_instance.best_solution()
```

```python
print(f"Najlepsze znalezione rozwiązanie (genom): {solution}")
print(f"Wartość funkcji przystosowania: {solution_fitness}")
```

Weryfikacja wyniku

```python
calculated_output = np.sum(solution * function_inputs)
```

```python
calculated_output.item()
```

Wykres zbieżności algorytmu

```python
ga_instance.plot_fitness(title="Zbieżność algorytmu PyGAD", save_dir=None)
```

### Przykład 3: partycjonowanie tablic

Na wejściu dostajemy wektor wartości liczbowych np. $[2, 10, 3, 8, 5, 7, 9, 5, 3, 2]$. Celem jest wyznaczenie takiego podziału wektora wejściowego, aby suma w dwóch podwektorach była równa, np. $[2, 10, 3, 5, 7]$ oraz $[8, 9, 5, 3, 2]$.

Określenie listy (tablicy) do podziału

```python
input_values = [2, 10, 3, 8, 5, 7, 9, 5, 3, 2]
```

```python
input_sum = np.sum(input_values)
```

Definicja funkcji podziału oraz przystosowania

```python
def binary_split(values: Iterable, binary_pos: Iterable) -> np.ndarray:
    subvec = np.array(values)[np.array(binary_pos) == 1]

    return subvec
```

```python
def fitness_func(ga_instance: Any, solution: Iterable, solution_idx: Any) -> float:
    subvec = binary_split(input_values, solution)
    subvec_sum = np.sum(subvec)

    # normalizacja wartości do zakresu 0-1
    fitness = 1 - np.abs(input_sum - 2 * subvec_sum) / input_sum

    return fitness.item()
```

Konfiguracja heurystyki

```python
ga_instance = pygad.GA(
    num_generations=30,
    gene_space=[0, 1],               # przestrzeń wartości genów - binarne
    fitness_func=fitness_func,       # zdefiniowana funkcja celu
    sol_per_pop=10,                  # rozmiar populacji (stała liczba osobników)
    num_genes=len(input_values),     # długość genomu (liczba zmiennych)
    num_parents_mating=5,            # liczba par rodziców
    keep_parents=2,                  # liczba rozwiązań do zachowania
    parent_selection_type="tournament", # selekcja turniejowa
    K_tournament=3,                  # rozmiar turnieju
    crossover_type="single_point",   # krzyżowanie jednopunktowe
    mutation_type="random",          # mutacja losowa
    mutation_percent_genes=10,        # szansa mutacji dla genu (%)
    stop_criteria="reach_1",         # zatrzymaj poszukiwania, gdy funkcja fitness osiągnie wartość 1
)
```

Uruchomienie ewolucji

```python
ga_instance.run()
```

Uzyskanie najlepszych wyników

```python
solution, solution_fitness, solution_idx = ga_instance.best_solution()
```

```python
solution, solution_fitness.item(), solution_idx.item()
```

### Przykład 4: problem plecaka

Plecak (ang. knapsack problem) to jeden z najbardziej klasycznych przykładów problemów optymalizacyjnych, którego celem jest wypełnienie plecaka (o ograniczonej nośności) przedmiotami, które osiągną łącznie jak największą wartość.

Przykładowo, pula przedmiotów:
1. Kurtka. Waga: 2 kg, wartość: 4$.
2. Śpiwór. Waga: 3 kg, wartość: 5$.
3. Aparat. Waga: 1 kg, wartość 3$.
4. Namiot. Waga: 4 kg, wartość: 7$.

Maksymalny udźwig plecaka: 5 kg. Co zatem najbardziej warto zabrać? Aparat (1 kg) i namiot (4 kg), co daje łączną wartość 10$.

```python
# klucz, [przedmiot, wartość, waga]
items = {
    1: ["Agregat", 100, 7],
    2: ["System HiFi", 300, 7],
    3: ["Śpiwór", 300, 10],
    4: ["Laptop", 500, 5],
    5: ["Dekoder TV", 200, 6],
    6: ["Radio", 40, 2],
    7: ["Aparat tlenowy", 300, 15],
    8: ["Ładowarka", 70, 6],
    9: ["Kubek termiczny", 250, 3],
    10: ["Namiot", 280, 3],
    11: ["Termos", 100, 1],
}
```

```python
max_weight = 25
```

```python
value_sum = sum([x[1] for x in items.values()])
```

Łączna wartość przedmiotów: 2440, łączna waga: 65

Definicja funkcji dopasowania

```python
def fitness_func(ga_instance: Any, solution: Iterable, solution_idx: Any) -> float:
    total_value = 0
    total_weight = 0

    for i in range(len(items)):
        if solution[i] == 0:
            continue

        total_value += items[i + 1][1]
        total_weight += items[i + 1][2]

        # ograniczenie wagi
        if total_weight > 25:
            return float("-inf")

    return total_value / value_sum
```

Konfiguracja heurystyki

```python
ga_instance = pygad.GA(
    num_generations=100,
    gene_space=[0, 1],
    num_parents_mating=15,
    fitness_func=fitness_func,
    sol_per_pop=40,
    num_genes=11,
    parent_selection_type="tournament",
    crossover_type="single_point",
    mutation_type="random",
    mutation_probability=0.02,
)
```

Uruchomienie ewolucji

```python
ga_instance.run()
```

Rozwiązania

```python
solution, solution_fitness, solution_idx = ga_instance.best_solution()
```
