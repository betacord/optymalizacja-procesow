# Wstępne przetwarzanie danych

## Regresja liniowa

W każdym zadaniu należy wykorzystać wszystkie poznane warianty regularyzacji modelu regresji liniowej.

1. Zadaniem jest przygotowanie danych do przewidywania jakości wina na podstawie jego właściwości fizykochemicznych (wine dataset). W tym celu należy: 
   1. wczytać zbiór danych do obiektu `DataFrame`,
   2. wybrać atrybut `quality` jako atrybut decyzyjny, a pozostałe kolumny wykorzystać jako zbiór atrybutów warunkowych, pomijając atrybut ID, który pełni jedynie funkcję identyfikatora obiektów,
   3. podzielić dane na podzbiór treningowy i testowy w proporcji 80:20, wykorzystując funkcję `train_test_split`,
   4. wyświetlić rozmiary otrzymanych podzbiorów i sprawdźić, czy liczba obiektów przed i po podziale jest zgodna.
   5. Wytrenować model na podzbiorze treningowym i zweryfikować błąd średniokwadratowy na podzbiorze testowym.
2. Celem zadania jest przewidywanie mediany wartości nieruchomości (`MedHouseVal`, zbiór California Housing) na podstawie pozostałych atrybutów. W tym celu należy: 
   1. wczytać zbiór danych i przygotuj `X` oraz `y`,
   2. wybrać kilka atrybutów numerycznych, np. `MedInc`, `AveRooms` i `Population`,
   3. podzielić zbiór danych na część treningową i testową,
   4. wykorzystać klasę `SimpleImputer` do uzupełnienia brakujących danych (jeśli istnieją, a jak nie to odrobinę temu dopomóc...) średnią arytmetyczną,
   5. wytrenować modele regresyjne na uzyskanym podzbiorze treningowym,
   6. powtórzyć eksperyment, wykorzystując medianę zamiast średniej i porównać wyniki obu podejść. Dlaczego wybór sposobu uzupełniania braków może wpływać na jakość modelu.
3. Celem zadania jest przewidywanie liczby wypożyczeń (atrybut `cnt`) na podstawie warunków pogodowych, daty i pozostałych informacji (zbiór [Bike Sharing Dataset](https://archive.ics.uci.edu/dataset/275/bike+sharing+dataset) zawierający informacje o dziennym wypożyczeniu rowerów - atrybuty liczbowe, jak i kategorialne). W tym celu należy:
   1. Wczytać dane i przygotować `X` oraz `y`,
   2. podzielić dane na podzbiór treningowy i testowy,
   3. wybrać atrybuty liczbowe i zastosować do nich dwie różne metody skalowania,
   4. wybrać atrybuty kategorialne i zastosować OneHotEncoder do zamiany atrybutów kategorialnych na postać numeryczną,
   5. wytrenować modele regresyjne,
   6. porównać wyniki uzyskane przy wykorzystaniu `StandardScaler` i `MinMaxScaler`,
   7. czy wszystkie atrybuty kategorialne powinny być kodowane za pomocą `OneHotEncoder`, czy w niektórych przypadkach uzasadnione byłoby użycie `OrdinalEncoder`,
   8. jak zmieniają się wyniki modeli L1, L2 i ElasticNet po zastosowaniu skalowania?
4. (*) Celem zadania jest przygotowanie kompletnego procesu przetwarzania danych i trenowania modelu z wykorzystaniem klas `Pipeline` oraz `ColumnTransformer` na przykładzie zbioru danych [Used Cars](https://www.kaggle.com/datasets/austinreese/craigslist-carstrucks-data). Wszystkie kroki powinny być wykonane analogicznie do zadań 1-3.

## Regresja logistyczna

W każdym zadaniu należy wykorzystać wszystkie poznane warianty regularyzacji modelu regresji logistycznej.

5. Celem zadania jest utworzenie modelu decyzyjnego, który na podstawie zbioru danych [KDD Cup 1999](https://archive.ics.uci.edu/dataset/130/kdd+cup+1999+data) (zawiera informacje o ruchu sieciowym) będzie rozpoznawał, czy obserwowana transmisja jest normalna, czy stanowi atak. W tym celu należy:
   1. wczytać zbiór danych za pomocą funkcji `fetch_kddcup99(as_frame=True)` - biblioteka `scikit-learn`,
   2. wybrać atrybut labels jako decyzyjny,
   3. przetransformować problem wieloklasowy w binarny (klasa 0 - ruch normalny, klasa 1 - dowolny atak),
   4. pozostałe kolumny wykorzystać jako warunkowe,
   5. podzielić zbiór na podzbiory treningowe i testowe w proporcji 80:20,
   6. wykorzystać parametr `stratify`, aby zachować proporcje obu klas w podzbiorze treningowym i testowym,
   7. wyświetlić liczbę obserwacji każdej klasy w obu podzbiorach,
   8. wytrenować model decyzyjny regresji logistycznej i sprawdzić skuteczność na zbiorze testowym za pomocą wskaźnika `accuracy`.
6. W tym zadaniu należy założyć, że w zbiorze danych mogą występować dane brakujące (jeśli nie - trzeba temu dopomóc poznanymi podejściami). Celem zadania jest porównanie wyników uzyskanych przez klasyfikatory z zadania 4 (na zbiorze oryginalnie wypełnionym danymi) ze zbiorem, gdzie dane wybrakowane (uprzednio dodane) zostały uzupełnione. Do porównania wykorzystać wskaźnik `accuracy` zarówno dla danych wybrakowanych (numerycznych) wypełnionych średnią arytemtyczną, oraz medianą.
7. Celem zadania będzie weryfikacja metod transformacji wartości numerycznych na zbiorze KDD Cup 1999. W tym celu należy:
   1. przygotować zbiór zbiór liczbowych i zastosować na nim poznane metody skalowania
   2. wybrać atrybuty kategorialne i wykorzystać klasę `OrdinalEncoder`, aby zakodować wybrane wartości kategorialne,
   3. powtórzyć eksperyment z punkt 2 z klasą `OneHotEncoder` jako osobny wariant zbioru danych,
   4. wytrenowac modele regresji logistycznej na obu wariantach zbioru danych,
   5. porównać skuteczność modeli wykorzystujących oba rodzaje kodowania za pomocą wskaźnika `accuracy`,
   6. czy `OrdinalEncoder` jest dobrym rozwiązaniem dla atrybutów takich jak `protocol_type` i `service`. Czy przypisanie kategoriom kolejnych liczb nie wprowadza przypadkiem nieistniejącej relacji porządkowej?
8. (*) Celem zadania jest opracowanie modelu decyzyjnego, przewidującego czy czy klient zrezygnuje z usług firmy telekomunikacyjnej za pomocą zbioru [Customer Churn](https://www.kaggle.com/datasets/muhammadshahidazeem/customer-churn-dataset) - dostępny w folderze `data`. Zadanie nalezy wykonać jako pełny potok przetwarzający dane - z wykorzystaniem klas `Pipeline` oraz `ColumnTransformer`. Wszystkie kroki powinny być wykonane analogicznie do zadań 5-7.
