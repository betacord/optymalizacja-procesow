# Optymalizacja gradientowa

## Optymalizacja funkcji wieloparametrycznych

1. Wykorzystać metodę optymalizacji gradientowej do znalezienia minimum funkcji $f(x)=(x-4)^2+3$ przy pochodnej równej $f'(x)=2(x-4)$. Przyjąć punkt startowy $x_0=0$, oraz 100 iteracji (parametr [options](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.minimize.html) - klucz maxiter).
2. Wykorzystać metodę optymalizacji gradientowej do znalezienia minimum funkcji $f(x)=x^4-4x^2+5$, której pochodna wynosi $f'(x)=4x^3-8x$. Przeprowadzić optymalizację dla trzech punktów początkowych: $x_0=-2$, $x_0=0.5$ oraz $x_0=2$. 
3. Pewna firma z Gdańska chce wyznaczyć wielkość produkcji $x$, która minimalizuje uproszczony model kosztu $C(x)=x^4-4x^2+px+10$, gdzie $p$ jest parametrem określającym wpływ dodatkowego kosztu jednostkowego. Należy przyjąc kolejno: $p=0$, $p=1$ oraz $p=3$. Pochodna funkcji kosztu wynosi: $C'(x)=4x^3-8x+p$. Celem zadania jest optymalizacja dla każdej z trzech wartości parametru $p$, porównanie znalezionych minimów i minimalnych kosztów, oraz weryfikacja czy rozwiązanie zależy od punktu startowego, i wyjaśnienie, jak zmiana parametru wpływa na ostateczny wynik. Czy dowolna wartość $x$ ma sens jako wielkość produkcji?

5. Na bazie zadania 4, przeprowadzić eksperymenty stosując pięć wartości współczynnika uczenia: $\eta\in\{0.00001, 0.0001, 0.001,\ 0.01,\ 0.1\}$. Dlaczego należy rozpoczynać każdy eksperyment od nowych parametrów modelu?
