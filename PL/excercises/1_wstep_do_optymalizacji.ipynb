# Wstęp do optymalizacji

## Optymalizacja jenowymiarowa bez ograniczeń

1. Znaleźć z wykorzystaniem `scipy.optimize.minimize_scalar` i metodą `bounded` minimum funkcji określonej wzorem $f(x) = x^4 - 3x^3 + 2x^2 + 5$ dla $x \in [-1, 3]$.
2. Opracować własną funkcję w Pythonie realizującą algorytm [Złotego Podziału](https://en.wikipedia.org/wiki/Golden-section_search) do wyznaczania minimum funkcji jednowymiarowej. Funkcja powinna przyjmować: funkcję celu, początkowy przedział $[a, b]$ oraz tolerancję $\epsilon$. Porównać liczbę iteracji własnej implementacji z funkcją `minimize_scalar` z biblioteki `scipy`.

## Optymalizacja wielowymiarowa bez ograniczeń
3. **Funkcja Rosenbrocka** jest klasycznym testem dla algorytmów optymalizacji: $f(x, y) = 100(y - x^2)^2 + (1 - x)^2$. Wykorzystać funkcję `scipy.optimize.minimize` z algorytmem bezgradientowym **Nelder-Mead**. Przetestować zbieżność startując z punktu $(-1.2, 1.0)$. Sprawdzić, czy algorytm osiągnął znane minimum globalne w punkcie $(1, 1)$.
4. Dana jest funkcja wielokrotnych minimów lokalnych: $f(x, y) = 20 + x^2 - 10 cos(2 \pi x) + y^2 - 10 cos(2 \pi y)$. Uruchomić optymalizację (funkcja `minimize` z metodą `BFGS`) dla 10 wygenerowanych punktów startowych metodą chybił-trafił z przedziału $[-5, 5]$. Zapisać uzyskane minima i wskazać, ile razy algorytm utknął w minimum lokalnym zamiast trafić w punkt $(0, 0)$.

## Optymalizacja z ograniczeniami
5. Zaprojektować walcowy zbiornik o objętości $V = 1000 cm^3$ tak, aby zminimalizowac jego pole powierzchni całkowitej $S(r, h) = 2 \pi r^2 + 2 \pi rh$. Przyjąć ograniczenia $V(r, h) = \pi r^2 h = 1000$ oraz $r > 0, h > 0$. Użyć funkcji `scipy.optimize.minimize` z metodą `SLSQP` i zdefiniować ograniczenie równościowe (parametr funkcji `constraints`, type: `eq`): https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.minimize.html.
