# Modelowanie matematyczne

## Zadanie 1: Inwestycje z limitem ryzyka całkowitego
Inwestor dysponuje kwotą $K = 500\,000\text{ PLN}$. Do wyboru ma 4 opcje inwestycyjne o rocznych stopach zwrotu:
* $R_1 = 4\%$ (niskie ryzyko)
* $R_2 = 7\%$ (średnie ryzyko)
* $R_3 = 12\%$ (wysokie ryzyko)
* $R_4 = 18\%$ (bardzo wysokie ryzyko)

Ograniczenia:
* Łączna kwota przeznaczona na opcje o wysokim i bardzo wysokim ryzyku ($R_3$ oraz $R_4$) nie może przekroczyć $15\%$ całego kapitału.
* Kwota w opcji $R_2$ musi stanowić co najmniej $25\%$ całego kapitału.
* Na opcję $R_4$ można przeznaczyć maksymalnie $5\%$ kapitału.

**Polecenie:**
Sformułuj szczegółowy model matematyczny (zmienne, parametry, ograniczenia, funkcja celu) w wersji pełnej oraz w (opcjonalnie) wersji skróconej z użyciem indeksów.

## Zadanie 2: Koszty stałe maszyn i stałe koszty uruchomienia
Firma produkuje $N = 2500$ detalicznych elementów na 3 maszynach. Koszt produkcji $P_i$ elementów na maszynie $i$ wynosi:
* $C_1 = 0.05 P_1^2 + 1.2 P_1 + 100 \cdot \beta_1$
* $C_2 = 0.8 P_2 + 250 \cdot \beta_2$
* $C_3 = 0.002 P_3^3 + 50 \cdot \beta_3$

Gdzie $\beta_i \in \{0, 1\}$ oznacza, czy $i$-ta maszyna została włączona. 

**Polecenie:**
Zapisz model matematyczny minimalizacji sumarycznego kosztu produkcji.

## Zadanie 3: Analiza wrażliwości i błędu reprezentacji (Refleksja)
W wykładzie podano definicję: *„Model matematyczny to funkcja, która przybliża rzeczywistość z pewnym znanym błędem”*.

**Polecenie:**
Na przykładzie problemu inwestycyjnego z Zadania 1 wskaż co najmniej **3** czynniki ze świata rzeczywistego, które powodują, że stworzony model jest jedynie przybliżeniem rzeczywistości (generują błąd reprezentacji).
