# Programowanie matematyczne

## Programowanie liniowe

1. Znaleźć wartości ciągłych parametrów $x,y$, które maksymalizują funkcję $2x+y$ przy ograniczeniach $-x+2y\leq 9,\quad 3x+y\leq18,\quad x-y\leq4$ oraz $0 \leq x\leq 10\, 0\leq y\leq10$. **Do wykonania** formalizacja i implementacja modelu matematycznewgo, oraz optymalizacja solverem liniowym i wypisanie $x, y$ oraz wartość funkcji celu.
2. Znaleźć wartości ciągłych parametrów $x,y$, które maksymalizują funkcję $3x + 2y$ przy ograniczeniach $-x+2y\leq8,\quad 2x+y\leq14,\quad 2x-y\leq10,\quad x+y\leq9$ oraz $0\leq x,y\leq10$. **Do wykonania** formalizacja i implementacja modelu matematycznewgo, oraz optymalizacja solverem liniowym i wypisanie $x, y$ oraz wartość funkcji celu.
3. Zakład wybiera wielkości produkcji dwóch wyrobów, oznaczone przez ciągłe parametry skalarne $x$ i $y$. Zysk wynosi 40 zł za jednostkę $x$ oraz 30 zł za jednostkę $y$. Dostępne zasoby opisują ograniczenia $2x+y\leq 100,\qquad x+2y\leq 80,\qquad x+y\leq 60$ oraz $x \geq 0 i y\geq 0$. **Do wykonania**: maksymalizacja funkcji $40x+30y$ oraz interpretacja, które ograniczenie zasobów ogranicza produkcję w znalezionym optimum.
4. (*) Proces ma trzy ciągłe parametry decyzyjne $x$, $y$ i $z$. Korzyść z procesu opisuje funkcja $f(x,y,z)=8x+7y+5z$. Ograniczenia zasobów są następujące: $x+2y+z\leq 24,\qquad 3x+y+2z\leq 30,\qquad x+y+z\leq 15$ oraz $x\geq 0,\qquad y\geq 0,\qquad z\geq 0$. **Do wykonania**: zapis pełnego model LP w `pyomo`. Po rozwiązaniu interpretacja, jak ograniczenia zasobów wpływają na kompromis między trzema parametrami.

## Programowanie całkowitoliczbowe

5. Zmaksymalizować funkcję $f(x,y)=5x+4y$ przy ograniczeniach $2x+y\leq 8,\qquad x+2y\leq 8$ oraz $x,y\in\mathbb{Z}_{\geq 0}$. **Do wykonania**: zdefiniować `model.x` i `model.y` jako parametry nieujemne typu całkowitego. Zoptymalizować model i porównać optimum z rozwiązaniem, które wyszłoby po dopuszczeniu wartości niecałkowitych.
6. Zmaksymalizować funkcję $f(x,y)=6x+5y$ przy ograniczeniach $2x+y\leq 10,\qquad x+2y\leq 9,\qquad x+y\leq 7$ oraz $x,y\in\mathbb{Z}_{\geq 0}$. **Do wykonania**: dodać wszystkie ograniczenia do modelu. Porównać wynik z zadaniem 5 i krótko wyjaśnić, dlaczego nie należy zakładać, że optimum całkowitoliczbowe zawsze pokryje się z optimum modelu ciągłego.
7. Zakład produkuje dwa wyroby w całkowitych partiach $x$ i $y$. Zysk wynosi odpowiednio 8 i 5 jednostek pieniężnych za partię. Ograniczenia dostępnych zasobów mają postać: $2x+3y\leq 20,\qquad 3x+y\leq 18$. Aby plan produkcji był opłacalny organizacyjnie, trzeba wyprodukować łącznie co najmniej 5 partii: $x+y\geq 5$, gdzie $x,y\in\mathbb{Z}_{\geq 0}$. **Do wykonania**: maksymalizacja funkcji $8x+5y$ oraz weryfikacja, czy dodatkowy warunek minimalnej skali zmienia rozwiązanie względem modelu bez tego warunku.
8. (*) Zakład wybiera całkowite wielkości produkcji $x$ i $y$. Dodatkowo parametr binarny $u$ oznacza, czy uruchomiono proces wytwarzający produkt \(x\). Uruchomienie kosztuje 10 jednostek pieniężnych. Nalezy zmaksymalizować $f(x,y,u)=8x+6y-10u$ przy ograniczeniach: $2x+3y\leq 24,\qquad x+2y\leq 16$, $3u\leq x\leq 7u,\qquad y\geq 2$, oraz $x,y\in\mathbb{Z}_{\geq 0},\qquad u\in\{0,1\}$. **Do wykonania**: model całkowitoliczbowy. Dodatkowo, wyjaśnić, w jaki sposób ograniczenie $3u\leq x\leq 7u$ wiąże decyzję o uruchomieniu procesu ze skalą produkcji.

## Mieszane programowanie liniowo-całkowitoliczbowe

9. Zmaksymalizować funkcję $f(x,y)=3x+2y$ przy ograniczeniach $2x+y\leq 16,\qquad x+y\leq 10$ oraz $0\leq x\leq 8,\qquad y\geq 0$, gdzie $x$ jest całkowite, a $y$ ciągłe. **Do wykonania**: utworzenie i optymalizacja modelu.
10. Zoptymalizować model z zadania 9 po dodaniu ograniczenia $x+2y\leq 14$. Pozostałe warunki i funkcja celu pozostają bez zmian. **Do wykonania**: porównanie wyniku z zadaniem 9.
11. W procesie produkcyjnym $x$ oznacza liczbę pełnych partii, natomiast $y$ oznacza ciągłą ilość dodatkowego surowca przetwarzanego niezależnie od pełnych partii. Zmaksymalizować $f(x,y)=9x+5y$ przy ograniczeniach $2x+y\leq 18,\qquad x+3y\leq 21$ oraz $0\leq x\leq 7,\qquad y\geq 0$, gdzie $x$ jest całkowite, a $y$ ciągłe. **Do wykonania**: optymalizacja modelu oraz porównanie wyniku z rozwiązaniem, w którym oba parametry byłyby ciągłe.
12. (*) Proces ma dwie ciągłe wielkości wyjściowe $x$ i $y$. Parametr binarny $u$ informuje, czy uruchomiono linię wytwarzającą $x$. Uruchomienie linii kosztuje 18 jednostek pieniężnych. Zmaksymalizować $f(x,y,u)=8x+5y-18u$ przy ograniczeniach $x+y\leq 12,\qquad x\leq 10u,\qquad x\geq 3u$ oraz $0\leq y\leq 6,\qquad x\geq 0,\qquad u\in\{0,1\}$. **Do wykonania**: optymalizacja modelu i interpretacja przypadku, kiedy model wybiera uruchomienie linii mimo kosztu stałego.

## Programowanie nieliniowe

13. Zmaksymalizować $f(x,y)=x+0.8xy$ przy ograniczeniach $-x+1.5xy\leq 7,\qquad 2x+y\leq 14,\qquad 2x-y\leq 10$ oraz $0\leq x\leq 10,\qquad 0\leq y\leq 10$. **Do wykonania**: utworzenie i optymalizacja modelu i wypisanie wyników.
14. Zmaksymalizować $f(x,y)=x+xy$ przy ograniczeniach $-x+2xy\leq 8,\quad 2x+y\leq 14,\quad 2x-y\leq 10$ oraz $x^2+y\leq 30,\qquad 0\leq x\leq 10,\qquad 0\leq y\leq 10$. **Do wykonania**: utworzenie i optymalizacja modelu. Dodatkowo, wskazać które elementy funkcji celu i ograniczeń są nieliniowe.
15. Jakość procesu zależy od dwóch ciągłych parametrów $x$ i $y$: $Q(x,y)=12x+10y-0.5x^2-0.4y^2+0.2xy$. Parametry muszą spełniać: $x+2y\leq 16,\qquad 3x+y\leq 18$ oraz $0\leq x\leq 10,\qquad 0\leq y\leq 8$. **Do wykonania**: maksymalizować $Q$. Dodatkowo, wyjaśnić, dlaczego model jest nieliniowy, chociaż oba ograniczenia zasobów są liniowe.
16. (*) Proces ma trzy ciągłe parametry: $x, y, z$. Korzyść opisuje funkcja $F(x,y,z)=16x+14y+10z-x^2-y^2-z^2+0.5xy$, a ograniczenia zasobów i jakości są następujące: $x+2y+z\leq 18,\qquad 2x+y+2z\leq 20,\qquad xy\geq 12$ oraz $0\leq x\leq 8,\qquad 0\leq y\leq 8,\qquad 0\leq z\leq 6$. **Do wykonania**: maksymalizacja $F$.
