# Metaheurystyki

## Maksymalizacja funkcji dwuparametrycznej

1. Dana jest funkcja $f(x, y) = sin(x) cos(y) + e^{- \frac{x^2 + y^2}{10}}$. Zmodyfikować przykład 1 z wykładu, tak aby algorytm genetyczny poszukiwał maksimum tej funkcji w zakresie $x,y\in[-3,3]$ zamiast $[-5,5]$. Wyświetlić na końcu współrzędne najlepszego znalezionego rozwiązania oraz wartość funkcji w tym punkcie.
2. Zmodyfikować funkcję celu, zastępując wykładnik w drugim składniku i doprowadzając w ten sposób funkcję do postaci $f(x, y) = sin(x) cos(y) + e^{- \frac{x^2 + y^2}{20}}$. Wyświetlić na końcu współrzędne najlepszego znalezionego rozwiązania oraz wartość funkcji w tym punkcie.
3. Znaleźć maksimum oryginalnej funkcji (z punktu 1), ale dodatkowo wymagając, aby rozwiązanie spełniało ograniczenie $x^2 + y^2 \leq 9$. Problem ten można rozwiązać na wiele sposobów i nie ma jednej dobrej odpowiedzi. Jedną z dróg może być modyfikacja funkcji przystosowania tak, aby rozwiązania niespełniające tego ograniczenia były nieatrakcyjne dla algorytmu. Można w tym celu zastosować odpowiednio dużą karę za naruszenie ograniczenia, czyli zwrócenie bardzo niskiej wartości przez funkcję przystosowania, gdy celem cały czas pozostaje znalezienie maksimum.
4. (*) Zakład produkuje dwa wyroby w ilościach $x$ i $y$. Zysk zależy od wielkości produkcji i opisuje go funkcja: $P(x, y) = 30x + 24y - 2x^2 - y^2 - xy$. Ilości wyrobów mogą przyjmować wartości rzeczywiste z przedziału $[0,10]$. Celem zadania jest implementacja zastosowania algorytmu genetycznego w bibliotece PyGAD, który znajdzie takie wartości $x$ i $y$, dla których zysk jest największy. Warto wykorzystać podejście z oryginalnego przykładu, opracowując jednocześnie nową funkcję celu i dostosowanie konfiguracji przestrzeni genów.

## Minimalizacja różnicy względem równania sześcioparametrowego

5. Dane jest równanie poszukujące wartości sześciu parametrów, dla których wyrażenie $11x_1+2.2x_2-0.7x_3+4x_4+8.9x_5-x_6$ będzie możliwie bliskie wartości $44$. Funkcja przystosowania premiuje rozwiązania o małym błędzie. Celem zadanie jest modyfikacja przykładu 2 z wykładu tak, aby algorytm poszukiwał rozwiązania, dla którego wartość wyrażenia liniowego będzie możliwie bliska 50 zamiast 44. Wyświetlić na końcu znalezione wartości wszystkich sześciu parametrów, obliczony wynik wyrażenia oraz bezwzględną różnicę między tym wynikiem a wartością docelową.
6. Dokonać modyfikacji równania, aby współczynnikb przy parametrze $x_5$ wynosił 9.1. Uruchomić algorytm genetyczny i sprawdzić, czy uzyskany wynik jest bliższy wartości docelowej niż wynik obliczony dla oryginalnego zestawu współczynników. Wyświetlić na końcu znalezione wartości wszystkich sześciu parametrów, obliczony wynik wyrażenia oraz bezwzględną różnicę między tym wynikiem a wartością docelową.
7. Zmodyfikować rozwiązanie zadania 5 tak, aby wszystkie sześć parametrów mogło przyjmować wyłącznie dodatnie wartości całkowite od $1$ do $10$. Uruchomić algorytm genetycznt, a następnie zweryfikować, czy znalezione rozwiązanie spełnia jednocześnie trzy wymagania: wszystkie parametrty są całkowite, wszystkie należą do dopuszczalnego przedziału, a wynik wyrażenia jest równy $44$ lub możliwie bliski tej wartości.
8. (*) Laboratorium opracowuje mieszaninę czterech składników. Ilości składników oznaczono jako $x_1,x_2,x_3,x_4$, a ich wkład do określonego wskaźnika jakościowego opisuje wyrażenie $Q=4x_1+7x_2+2x_3+5x_4$. Celem zadania jest dobranie ilości składników tak, aby wskaźnik $Q$ był możliwie bliski wartości $100$. Każda ilość może przyjmować dowolną wartość rzeczywistą z przedziału $[0,20]$. Po zakończeniu optymalizacji wyświetlić znalezione ilości składników, uzyskaną wartość $Q$ oraz błąd względem wartości docelowej. Sprawdź, czy wszystkie parametry spełniają ograniczenia.


## Partycjonowanie wektorów

9. Zmodyfikować przykład 3 z wykładu, tak by aby podzielił następujący zbiór na dwa podzbiory o możliwie równych sumach: $
[4,6,3,7,8,2]$. Po zakończeniu działania wypisać elementy obu podzbiorów, ich sumy oraz różnicę między tymi sumami.
10.  Zmodyfikować oryginalny przykład nr 3 z wykładu tak, aby celem nie było uzyskanie dwóch równolicznych sum, lecz znalezienie podzbioru, którego suma będzie możliwie bliska jednej trzeciej sumy wszystkich elementów, a pozostałe elementy mają tworzyć drugi podzbiór. Po zakończeniu działania wypisać oba podzbiory, sumę wybranego podzbioru oraz docelową wartość, do której dąży algorytm.
11. Rozszerzyć implementację problemu partycjonowania tak, aby oba podzbiory musiały spełniać jednocześnie dwa warunki: mieć jednakową sumę oraz liczbę elementów. W tym celu wykorzystać wektor $[3,5,6,8,9,11,12,14]$. Warto w tym celu zmodyfikować funkcję przystosowania tak, aby uwzględniała zarówno różnicę sum, jak i różnicę liczebności podzbiorów. Przygotować ocenę jakości rozwiązania uwzględniającą oba kryteria. Po zakończeniu optymalizacji przedstawić oba podzbiory i zweryfikować osobno każdy z wymaganych warunków.
12. (*) Zmodyfikować oryginalny przykład nr 3 z wykładu, tak aby dzielił wektor $[2,10,3,8,5,7,9,5,3,2]$ na trzy podzbiory o możliwie równych sumach. W tym przypadku każdy gen powinien wskazywać przynależność elementu do jednego z trzech podzbiorów. Warto zmienić w tym rozwiązaniu reprezentację genotypu oraz funkcję przystosowania tak, aby algorytm dążył do minimalizacji różnic między sumami trzech grup. Dodatkowo należy wymagać, aby żaden podzbiór nie był pusty. Po zakońxzeniu  optymalizacji przedstawić skład każdej grupy, jej sumę oraz największą różnicę między sumami grup.

## Problem plecakowy

13.  W przykładzie nr 4 z wykładu, każdy przedmiot jest reprezentowany przez gen binarny, gdzie 1 oznacza zabranie przedmiotu, a 0 — pozostawienie go. Celem zadania jest maksymalizacja wartości wybranych przedmiotów przy ograniczeniu łącznej wagi do 25 kg. W tym celu należy pozostawić oryginalną listę przedmiotów i konfigurację optymalizatora bez zmian. Wystarczy rozszerzyć jedynie część programu odpowiedzialną za prezentację wyniku. Po zakończeniu optymalizacji wyświetlić: nazwy wybranych przedmiotów, ich łączną wartośc, łączną wagę oraz pozostały udźwig plecaka.
14.  Rozszerzyć oryginalną listę o nowy przedmiot: powerbank o wartości 120 i wadze 2 KG. Pozostawić maksymalną wagę plecaka równą 25 kg. Dostosować program tak, aby nowy przedmiot również był uwzględniany przez algorytm genetyczny. Warto w tym celu zmodyfikować kod tak, aby liczba genów wynikała z liczby zdefiniowanych przedmiotów, zamiast być wpisana na stałe. Finalnie uruchomić optymalizację i porównać znaleziony zestaw z wynikiem uzyskanym przed dodaniem nowego przedmiotu.
15. Rozszerzyć problem plecakowy o dodatkowe wymaganie: można wybrać najwyżej cztery przedmioty. Proszę pozostawić oryginalny zbiór przedmiotów i limit wagowy 25 kg. W tym celu warto zmodyfikować funkcję przystosowania tak, aby uwzględniała oba ograniczenia: łączną wagę oraz liczbę wybranych przedmiotów. Po zakończeniu działania algorytmu zweryfikować spełnienie obu warunków. Finalnie przedstawić wybraną kombinację przedmiotów i jej łączną wartość, oraz porównać wynik z rozwiązaniem uzyskanym bez ograniczenia liczby przedmiotów.
16. (*) Firma dysponuje budżetem w wysokości 50 tys. zł i rozważa realizację siedmiu projektów. Każdy projekt charakteryzuje się kosztem oraz przewidywanym zyskiem, przedstawionym w tabeli ponizej. Zbudować algorytm genetyczny, który wybierze projekty zapewniające największy łączny zysk, przy czym suma ich kosztów nie może przekroczyć dostępnego budżetu. Warto Wykorzystać w tym celu binarną reprezentację genotypu podobnie jak w problemie plecakowym oraz dostosować dane wejściowe, funkcję przystosowania i konfigurację PyGAD do nowej sytuacji i opracować sposób obsługi rozwiązań przekraczających budżet.
