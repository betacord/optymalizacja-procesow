# Modelowanie matematyczne

Przed przystąpieniem do prac nad optymalizacją, konieczne jest zapoznanie się z koncepcją i pojęciami tematyki **modelowania matematycznego**, której rolą jest formalna reprezentacja procesów występujących w swiecie rzeczywistym, a także abstrakcyjnym. 

Samo pojęcie **modelu matematycznego** odnosi się do funkcji matematycznej, która przybliża rzeczywistość z pewnym, znanym błędem. Pojęcie to sprawnie łączy się z metodami optymalizacji, co prowadzi najczęściej do minimalizacji wspomnianego błędu reprezentacji rzeczywistości. Zastosowania optymalizacji w tym przypadku potrafią być jednak znacznie szersze.

## Rola modelowania matematycznego

W optymalizacji (w tym także data science, uczenie maszynowe, itd.) podczas realizacji każdego projektu, stosuje się 4-etapowy schemat prac, przeznaczony do skutecznego podziału zakresów poszczególnych kroków.

1. Zrozumienie problemu
2. Modelowanie problemu (etap tworzenia modelu matematycznego)
3. Rozwiązanie (metody, narzędzia i implementacja)
4. Analiza wyników

Modelowanie matematycznego dotyczy ściśle etapów 1 i 2, gdzie modelowanie problemu jest etapem prowadzącym do utworzenia modelu matematycznego. Sam model matematyczny jest elementem koniecznym do wdrożenia prac w etapie 3., a etap 4. polega na analizie wyników uzyskanych przez przyjęte rozwiązania. Nie sposób nie zauważyć, że model matematyczny jest najbardziej krytycznym etapem całego procesu, którego poprawna realizacja, warunkuje szanse na sukces całego procesu.

## Pojęcia

Zarówno w modelowaniu matematycznym, jak też w optymalizacji, ważne jest stosowanie spójnej terminologii.

1. Funkcja celu - funkcja, która będzie poddawana procesowi optymalizacji, czyli poszukiwania optymalnego rozwiązania.
2. Zmienna - wartość wyjściowa funkcji (modelu matematycznego), której optimum (minimum lub maksimum) będzie poszukiwane podczas procesu optymalizacji.
3. Parametr (najczęściej w liczbie mnogiej) - wartości wejściowe do funkcji, które definiują wartość zmiennej. Poszukiwanie optimum będzie polegało na dobieraniu wartości parametrów, które zapewnią optimum.
4. Ograniczenia (lub reguły) - matematycznie sformułowane reguły przybliżające obszar poszukiwań optimum.
5. Indeks - zmienne i parametry, z uwagi na wspólny symbol, są zazwyczaj dodatkowo indeksowane numerycznie celem jednoznacznej formalizacji we wzorach matematycznych.

## Przykłady

### 1. Inwestycje finansowe

Mamy do dyspozycji 100000 PLN i szukamy opcji inwestycyjnych maksymalizujących zysk. Mamy do dyspozycji następujące rozwiązania:

1. inwestycja o niskim ryzyku straty z roczną stopą zwrotu w wysokości 5%,
2. inwestycja o średnim ryzyku straty z roczną stopą zwrotu w wysokości 10%,
3. inwestycja o wysokim ryzyku straty z roczną stopą zwrotu w wysokości 15%.

Zdrowy rozsądek sugeruje nam jednak, aby nie przeznaczać wszystkich środków na inwestycje o średnim lub wysokim ryzyku utraty kapitału. Przeznaczmy maksymalnie 10% całego kapitału na inwestycję o wysokim ryzyku i maksymalnie 20% na inwestycję o średnim ryzyku. 

<ins>Jak sformułować model matematyczny?</ins>

**Zmienne**, czyli co przewidujemy
- $R_1$ - zysk z pierwszej opcji inwestycyjnej
- $R_2$ - zysk z drugiej opcji inwestycyjnej
- $R_3$ - zysk z trzeciej opcji inwestycyjnej

**Parametry**, czyli jak sterujemy przewidywaniem z zewnątrz
- $C_1$ - kwota przeznaczona na pierwszą opcję inwestycyjną
- $C_2$ - kwota przeznaczona na drugą opcję inwestycyjną
- $C_3$ - kwota przeznaczona na trzecią opcję inwestycyjną

**Ograniczenia**, czyli według jakich reguł postępuje model
- $C_1 + C_2 + C_3 = 100000$
- $R_1 = 0.05 C_1$
- $R_2 = 0.1 C_2$
- $R_3 = 0.15 C_3$
- $0 \leq C_2 \leq 0.2 \times 100000$
- $0 \leq C_3 \leq 0.1 \times 100000$

**Funkcja celu**
- $max(R_1 + R_2 + R_3)$



Całość można jednak zapisać sprawniej, unikając wielu nadmiarowych powtórzeń, w szczególności dla dużej liczby kryteriów

**Zbiór indeksów**
- $F = \{1, 2, 3\}$

**Ograniczenia**
- $\sum_{f \in F} C_f = 100000$
- $R_f = return(C_f) \forall f \in F$
- $C_f^{min} \leq C_f \leq C_f^{max} \forall f \in F$

**Funkcja celu**
- $max \sum_{f \in F} R_f$

### 2. Koszty produkcji stołów

Mamy do dyspozycji proces produkcyjny stołów kuchennych, do których części są produkowane przez trzy maszyny. Naszym zadaniem jest dobór liczby części dla każdej z maszyn w taki sposób, aby zminimalizować koszt produkcji, przy założeniu, że w kolejnym miesiącu celem jest wyprodukowanie 1000 stołów.

Koszt produkcji każdej maszyny można wyrazić następująco:
1. $C_1 = 0.1P_A^2 + 0.5P_A + 0.1$
2. $C_2 = 0.3P_B + 0.5$
3. $C_3 = 0.01 P_3^3$

gdzie $P_i$ oznacza liczbę produktów powstałych w $i$-tej maszynie.

**Zmienne**
- $C_1$ - koszt produkcji maszyny 1
- $C_2$ - koszt produkcji maszyny 2
- $C_3$ - koszt produkcji maszyny 3

**Parametry**
- $P_1$ - liczba produktów z maszyny 1
- $P_2$ - liczba produktów z maszyny 2
- $P_3$ - liczba produktów z maszyny 3

**Ograniczenia**
- $P_1 + P_2 + P_3 = 1000$
- $C_1 = 0.1P_1^2 + 0.5P_1 + 0.1$
- $C_2 = 0.3P_2 + 0.5$
- $C_3 = 0.01 P_3^3$
- $P_1, P_2, P_3 \geq 0$

**Funkcja celu**
- $min(C_1 + C_2 + C_3)$

Jak to wyrazić skrótowo i w formie bardziej uogólnionej?

**Zbiór indeksów**
- $F = \{1, 2, 3\}$

**Ograniczenia**
- $\sum_i P_i = 1000$
- $C_1 = 0.1 P_1^2 + 0.5 P_1 + \beta_1 0.1$
- $C_2 = 0.3P_2 + \beta_2 0.5$
- $C_3 = 0.01 P_3^3$
- $P_1 \leq \beta_1 M$
- $P_2 \leq \beta_2 M$
- $P_1, P_2, P_3 \geq 0$

**Funkcja celu**
- $min \sum_i C_i$

Dodatkowe symbole
- $M$ - dowolna bardzo duża liczba reprezentująca koszt samej pracy maszyny
- $\beta_j$ - zmienna binarna reprezentująca włączony stan maszyny

Warto zauważyć, że koszt pracy maszyn 1 i 2 zależy nie tylko od liczby wyprodukowanych części. Wyraz wolny wielomianu definiującego ten koszt jest dodatkowym kosztem samej pracy maszyny, bez względu na liczbę wyprodukowanych części. Przy zerowej produkcji, koszt ten także będzie ponoszony tylko ze względu na fakt, że maszyna jest włączona.
