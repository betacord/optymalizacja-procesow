## Unconstrained Multi-Variable Optimization
3. The **Rosenbrock function** is a classic test for optimization algorithms: $f(x, y) = 100(y - x^2)^2 + (1 - x)^2$. Use the `scipy.optimize.minimize` function with the **Nelder-Mead** derivative-free algorithm. Test convergence starting from the point $(-1.2, 1.0)$. Check whether the algorithm reached the known global minimum at the point $(1, 1)$.
4. Consider the function with multiple local minima: $f(x, y) = 20 + x^2 - 10 \cos(2 \pi x) + y^2 - 10 \cos(2 \pi y)$. Perform optimization (using the `minimize` function with the `BFGS` method) starting from 10 points generated randomly within the interval $[-5, 5]$. Record the resulting minima and indicate how many times the algorithm became stuck in a local minimum instead of reaching the point $(0, 0)$.
