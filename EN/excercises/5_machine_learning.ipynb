## Classification

5. In the lecture example using the `Breast Cancer Wisconsin` dataset, the data was split into training and test sets in an 80%:20% ratio. Compare the obtained results with the original example. Which model achieved the highest accuracy? Did changing the data split ratio affect the ranking of the models in terms of classification performance?
6. Using the `Breast Cancer Wisconsin` dataset, investigate the effect of regularization parameters on classification performance. Conduct an experiment using the following values of the regularization-strength hyperparameter: $[0.01, 0.1, 1.0, 10.0]$, with three variants: L1, L2, and ElasticNet (with `l1_ratio=0.5`). For each configuration: train the model, calculate the accuracy metric on the training and test subsets, display the model coefficient values, and count the number of non-zero coefficients. Based on the obtained results, answer the following questions: How does changing the hyperparameter affect regularization strength? Which method sets some coefficients to zero? For each configuration: predict the probability of belonging to the malignant class; apply the selected decision threshold to assign class labels; and calculate the accuracy metric. Finally, identify the model configuration and decision threshold that provide the highest accuracy.
