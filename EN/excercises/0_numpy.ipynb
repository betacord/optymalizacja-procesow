## Numpy

All files referenced in the exercises are located in the `data` directory.

1. Create an array containing fifty values equal to 5.
2. Create a 5×5 array containing the values from 1 to 25.
3. Create an array of even numbers from 10 to 50.
4. Create a matrix with 8 on the diagonal and 0 elsewhere.
5. Create a 10×10 array with values increasing by 0.01.
6. Create a linear space of 50 values in the range 0–1.
7. From the array created in Exercise 2, select a 12-element subarray containing values in the range 12–25.
8. Select the first 3 elements from the last column of the array created in Exercise 2 and arrange them as a column vector.
9. Compute the sum of the elements in the last two rows of the matrix created in Exercise 2.
10. Write a script that creates a tensor containing random integer values, with a random number of dimensions and a random size in each dimension.
11. Create a matrix in which every row contains the values 1 to 5. Do not use a ready-made list to construct the matrix.
12. Create a 10-element vector of random values and replace its smallest value with 0.
13. Write a script that finds, in a random vector `Z`, the value closest to a random scalar `v`.
14. Generate a random 5×8 matrix `M` and swap two randomly selected rows.
15. Generate a 5×5 matrix filled with any sequence of consecutive values and transform it into a column vector while preserving all values.
16. Create a 6×6 matrix composed of four 3×3 blocks:
- upper-left: all ones,
- upper-right: all twos,
- lower-left: all threes,
- lower-right: a diagonal matrix with 9 on the diagonal.
17. Set all elements with values > 2 to zero.
18. Generate a 30-element integer vector with values in the range [-20, 20]. Replace negative values by their absolute values, but only at positions with even indices. Finally, compute the sum of elements whose indices are in the interval 10–20 inclusive.
19. Load `iris.data` (comma-separated columns; the last column is a text label). Keep only rows for which sepal length > 6.0 and the species is `Iris-virginica`. From the filtered data, build a feature matrix containing the 4 numerical attributes after removing the label column, and compute the mean of each feature.
Load `seeds_dataset.txt` (space-separated values; the last column is class 1/2/3). Keep only records belonging to the selected class. In the resulting matrix, randomly choose two different row indices and swap those rows. Compute the sum of the elements in the last two rows.
20. Load `wine.data` (CSV; the first column is class 1/2/3 and the remaining columns are features). Select only columns 2, 5, and 10 and use them to build matrix `X`. Set to zero all values in `X` that are smaller than the median of `X` (with the median computed over all elements of `X`).
