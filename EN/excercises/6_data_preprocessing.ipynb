## Logistic regression

In each task, use all the learned regularization variants of the logistic regression model.

5. The goal of the task is to create a decision model that, based on the [KDD Cup 1999](https://archive.ics.uci.edu/dataset/130/kdd+cup+1999+data) dataset (which contains information about network traffic), will identify whether an observed transmission is normal or an attack. To do this:
   1. load the dataset using the `fetch_kddcup99(as_frame=True)` function from the `scikit-learn` library,
   2. select the `labels` attribute as the target attribute,
   3. transform the multiclass problem into a binary one (class 0 – normal traffic, class 1 – any attack),
   4. use the remaining columns as input features,
   5. split the dataset into training and test subsets in an 80:20 ratio,
   6. use the `stratify` parameter to preserve the proportions of both classes in the training and test subsets,
   7. display the number of observations from each class in both subsets,
   8. train a logistic regression decision model and check its performance on the test set using the `accuracy` metric.
6. In this task, assume that the dataset may contain missing data (if it does not, introduce some using the approaches learned). The goal of the task is to compare the results obtained by the classifiers from task 4 (on the dataset originally containing complete data) with the results obtained on a dataset in which missing data (added beforehand) have been imputed. Use the `accuracy` metric for comparison, considering missing numerical values imputed using both the arithmetic mean and the median.
7. The goal of this task is to verify methods for transforming numerical values on the KDD Cup 1999 dataset. To do this:
   1. prepare a set of numerical features and apply the learned scaling methods to it,
   2. select categorical attributes and use the `OrdinalEncoder` class to encode the selected categorical values,
   3. repeat the experiment from step 2 using the `OneHotEncoder` class as a separate variant of the dataset,
   4. train logistic regression models on both dataset variants,
   5. compare the performance of models using both encoding methods with the `accuracy` metric,
   6. is `OrdinalEncoder` a good solution for attributes such as `protocol_type` and `service`? Does assigning consecutive numbers to categories perhaps introduce a non-existent ordering relationship?
