# Mathematical modelling

## Exercise 1: Investments with a total-risk limit
An investor has capital of $K=500\,000\text{ PLN}$. Four investment options are available, with annual rates of return:
* $R_1=4\%$ (low risk)
* $R_2=7\%$ (medium risk)
* $R_3=12\%$ (high risk)
* $R_4=18\%$ (very high risk)

Constraints:
* The total amount allocated to the high- and very-high-risk options ($R_3$ and $R_4$) cannot exceed $15\%$ of the total capital.
* The amount invested in $R_2$ must be at least $25\%$ of the total capital.
* At most $5\%$ of the capital may be allocated to $R_4$.

**Task:**
Formulate a detailed mathematical model (variables, parameters, constraints, and objective function) in full form and, optionally, in a compact indexed form.

## Exercise 2: Machine fixed costs and start-up costs
A company produces $N=2500$ components using 3 machines. The cost of producing $P_i$ components on machine $i$ is:
* $C_1=0.05P_1^2+1.2P_1+100\cdot\beta_1$
* $C_2=0.8P_2+250\cdot\beta_2$
* $C_3=0.002P_3^3+50\cdot\beta_3$

where $\beta_i\in\{0,1\}$ indicates whether machine $i$ is switched on.

**Task:**
Write a mathematical model that minimises the total production cost.
