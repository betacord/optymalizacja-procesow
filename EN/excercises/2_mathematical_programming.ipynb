# Mathematical Programming

## Linear Programming

1. Find the values of the continuous variables $x,y$ that maximise the function $2x+y$ subject to the constraints $-x+2y\leq 9,\quad 3x+y\leq18,\quad x-y\leq4$ and $0 \leq x\leq 10,\quad 0\leq y\leq10$. **Tasks:** formulate and implement the mathematical model, optimise it using a linear programming solver, and report the values of $x, y$ and the objective function value.
2. Find the values of the continuous variables $x,y$ that maximise the function $3x + 2y$ subject to the constraints $-x+2y\leq8,\quad 2x+y\leq14,\quad 2x-y\leq10,\quad x+y\leq9$ and $0\leq x,y\leq10$. **Tasks:** formulate and implement the mathematical model, optimise it using a linear programming solver, and report the values of $x, y$ and the objective function value.
3. A factory must decide how many units of two products to manufacture, represented by the continuous scalar variables $x$ and $y$. The profit is PLN 40 per unit of $x$ and PLN 30 per unit of $y$. The available resources are described by the constraints $2x+y\leq 100,\qquad x+2y\leq 80,\qquad x+y\leq 60$ and $x \geq 0,\qquad y\geq 0$. **Tasks:** maximise the function $40x+30y$ and explain which resource constraint limits production at the optimum.
4. (*) A process has three continuous decision variables, $x$, $y$ and $z$. The benefit obtained from the process is described by the function $f(x,y,z)=8x+7y+5z$. The resource constraints are $x+2y+z\leq 24,\qquad 3x+y+2z\leq 30,\qquad x+y+z\leq 15$ and $x\geq 0,\qquad y\geq 0,\qquad z\geq 0$. **Tasks:** formulate the complete LP model in `pyomo`. After solving it, explain how the resource constraints affect the trade-off between the three decision variables.

## Integer Programming

5. Maximise the function $f(x,y)=5x+4y$ subject to the constraints $2x+y\leq 8,\qquad x+2y\leq 8$ and $x,y\in\mathbb{Z}_{\geq 0}$. **Tasks:** define `model.x` and `model.y` as non-negative integer variables. Optimise the model and compare the optimum with the solution obtained if non-integer values were allowed.
6. Maximise the function $f(x,y)=6x+5y$ subject to the constraints $2x+y\leq 10,\qquad x+2y\leq 9,\qquad x+y\leq 7$ and $x,y\in\mathbb{Z}_{\geq 0}$. **Tasks:** add all the constraints to the model. Compare the result with Exercise 5 and briefly explain why one should not assume that the integer optimum will always coincide with the optimum of the continuous model.
7. A factory produces two products in whole batches, represented by $x$ and $y$. The profit is 8 and 5 monetary units per batch, respectively. The available resource constraints are $2x+3y\leq 20,\qquad 3x+y\leq 18$. To make the production plan organisationally viable, at least 5 batches must be produced in total: $x+y\geq 5$, where $x,y\in\mathbb{Z}_{\geq 0}$. **Tasks:** maximise the function $8x+5y$ and determine whether the additional minimum-production requirement changes the solution compared with the model without this requirement.
8. (*) A factory chooses integer production quantities $x$ and $y$. An additional binary variable $u$ indicates whether the process producing product \(x\) has been started. Starting the process costs 10 monetary units. Maximise $f(x,y,u)=8x+6y-10u$ subject to the constraints $2x+3y\leq 24,\qquad x+2y\leq 16$, $3u\leq x\leq 7u,\qquad y\geq 2$, and $x,y\in\mathbb{Z}_{\geq 0},\qquad u\in\{0,1\}$. **Tasks:** formulate the integer programming model. Also explain how the constraint $3u\leq x\leq 7u$ links the decision to start the process to the production quantity.

## Mixed-Integer Linear Programming

9. Maximise the function $f(x,y)=3x+2y$ subject to the constraints $2x+y\leq 16,\qquad x+y\leq 10$ and $0\leq x\leq 8,\qquad y\geq 0$, where $x$ is an integer variable and $y$ is continuous. **Tasks:** formulate and optimise the model.
10. Optimise the model from Exercise 9 after adding the constraint $x+2y\leq 14$. All other constraints and the objective function remain unchanged. **Tasks:** compare the result with Exercise 9.
11. In a production process, $x$ represents the number of whole batches, while $y$ denotes a continuous quantity of additional raw material processed independently of those batches. Maximise $f(x,y)=9x+5y$ subject to the constraints $2x+y\leq 18,\qquad x+3y\leq 21$ and $0\leq x\leq 7,\qquad y\geq 0$, where $x$ is an integer variable and $y$ is continuous. **Tasks:** optimise the model and compare the result with the solution obtained when both variables are continuous.
12. (*) A process has two continuous output quantities, $x$ and $y$. The binary variable $u$ indicates whether the production line for $x$ has been started. Starting the line costs 18 monetary units. Maximise $f(x,y,u)=8x+5y-18u$ subject to the constraints $x+y\leq 12,\qquad x\leq 10u,\qquad x\geq 3u$ and $0\leq y\leq 6,\qquad x\geq 0,\qquad u\in\{0,1\}$. **Tasks:** optimise the model and interpret the case in which the model chooses to start the production line despite the fixed cost.

## Nonlinear Programming

13. Maximise $f(x,y)=x+0.8xy$ subject to the constraints $-x+1.5xy\leq 7,\qquad 2x+y\leq 14,\qquad 2x-y\leq 10$ and $0\leq x\leq 10,\qquad 0\leq y\leq 10$. **Tasks:** formulate and optimise the model, and report the results.
14. Maximise $f(x,y)=x+xy$ subject to the constraints $-x+2xy\leq 8,\quad 2x+y\leq 14,\quad 2x-y\leq 10$ and $x^2+y\leq 30,\qquad 0\leq x\leq 10,\qquad 0\leq y\leq 10$. **Tasks:** formulate and optimise the model. Also identify which terms in the objective function and constraints are nonlinear.
15. The quality of a process depends on two continuous variables, $x$ and $y$: $Q(x,y)=12x+10y-0.5x^2-0.4y^2+0.2xy$. The variables must satisfy the constraints $x+2y\leq 16,\qquad 3x+y\leq 18$ and $0\leq x\leq 10,\qquad 0\leq y\leq 8$. **Tasks:** maximise $Q$. Also explain why the model is nonlinear even though both resource constraints are linear.
16. (*) A process has three continuous variables: $x, y, z$. The benefit is described by the function $F(x,y,z)=16x+14y+10z-x^2-y^2-z^2+0.5xy$, and the resource and quality constraints are $x+2y+z\leq 18,\qquad 2x+y+2z\leq 20,\qquad xy\geq 12$ and $0\leq x\leq 8,\qquad 0\leq y\leq 8,\qquad 0\leq z\leq 6$. **Tasks:** maximise $F$.
