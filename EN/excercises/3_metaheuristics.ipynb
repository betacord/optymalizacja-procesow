## Two-Parameter Function Maximization

1. Given the function $f(x, y) = \sin(x) \cos(y) + e^{- \frac{x^2 + y^2}{10}}$. Modify Example 1 from the lecture so that the genetic algorithm searches for the maximum of this function in the range $x,y\in[-3,3]$ instead of $[-5,5]$. At the end, display the coordinates of the best solution found and the value of the function at that point.
2. Modify the objective function by replacing the exponent in the second term, thereby bringing the function to the form $f(x, y) = \sin(x) \cos(y) + e^{- \frac{x^2 + y^2}{20}}$. At the end, display the coordinates of the best solution found and the value of the function at that point.
3. Find the maximum of the original function (from point 1), but additionally requiring the solution to satisfy the constraint $x^2 + y^2 \leq 9$. This problem can be solved in many ways and there is no single correct answer. One approach could be to modify the fitness function so that solutions not meeting this constraint are unattractive to the algorithm. To achieve this, a sufficiently large penalty for constraint violation can be applied, i.e., returning a very low value by the fitness function while the objective remains finding the maximum.

The goal of the task is to implement the application of a genetic algorithm in the PyGAD library to find the values of $x$ and $y$ that yield the maximum profit. It is worth using the approach from the original example while developing a new objective function and adjusting the gene space configuration.

## Minimization of Difference Relative to a Six-Parameter Equation

The fitness function rewards solutions with a small error. The goal of the task is to modify Example 2 from the lecture so that the algorithm searches for a solution where the value of the linear expression is as close as possible to 50 instead of 44. At the end, display the found values of all six parameters, the calculated result of the expression, and the absolute difference between this result and the target value.
6. Modify the equation so that the coefficient for parameter $x_5$ is 9.1. Run the genetic algorithm and check whether the obtained result is closer to the target value than the result calculated for the original set of coefficients. At the end, display the found values of all six parameters, the calculated result of the expression, and the absolute difference between this result and the target value.
7. Modify the solution to task 5 so that all six parameters can take only positive integer values from $1$ to $10$. Run the genetic algorithm, and then verify whether the found solution simultaneously satisfies three requirements: all parameters are integers, all belong to the allowable range, and the result of the expression is equal to $44$ or as close as possible to that value.

## Knapsack Problem

13. In Example 4 from the lecture, each item is represented by a binary gene, where 1 means taking the item and 0 means leaving it. The goal of the task is to maximize the value of selected items with a total weight limit of 25 kg. For this purpose, leave the original item list and optimizer configuration unchanged. It is sufficient to extend only the part of the program responsible for presenting the result. After optimization, display: the names of the selected items, their total value, total weight, and the remaining capacity of the knapsack.
14. Extend the original list with a new item: a power bank with a value of 120 and a weight of 2 kg. Leave the maximum knapsack weight at 25 kg. Adjust the program so that the new item is also considered by the genetic algorithm. To do this, modify the code so that the number of genes is derived from the number of defined items rather than being hardcoded. Finally, run the optimization and compare the found set with the result obtained before adding the new item.
15. Extend the knapsack problem with an additional requirement: at most four items can be selected. Please keep the original set of items and the 25 kg weight limit. Finally, present the selected combination of items and its total value, and compare the result with the solution obtained without the item count limit.
