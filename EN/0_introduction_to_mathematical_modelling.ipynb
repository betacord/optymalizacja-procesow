# Mathematical modelling

Before starting work on optimisation, it is necessary to become familiar with the concept and terminology of **mathematical modelling**, whose role is to provide a formal representation of processes occurring in the real world as well as in abstract settings.

The term **mathematical model** itself refers to a mathematical function that approximates reality with a certain known error. This concept combines naturally with optimisation methods, which most often leads to minimising the aforementioned error in the representation of reality. However, the applications of optimisation in this context can be much broader.

## The role of mathematical modelling

In optimisation (including data science, machine learning, etc.), a four-stage workflow is commonly used when carrying out a project. It helps divide the work into well-defined stages.

1. Understanding the problem
2. Modelling the problem (creating a mathematical model)
3. Solving the problem (methods, tools, and implementation)
4. Analysing the results

Mathematical modelling is closely related to stages 1 and 2, where problem modelling leads to the construction of a mathematical model. The mathematical model itself is necessary in order to proceed with stage 3, while stage 4 consists of analysing the results obtained using the adopted solution. It is therefore clear that the mathematical model is one of the most critical elements of the entire process: if it is constructed correctly, the chances of success of the whole project increase substantially.

## Terminology

In both mathematical modelling and optimisation, it is important to use consistent terminology.

1. Objective function – the function that is subject to optimisation, i.e. the search for an optimal solution.
2. Variable – an output value of a function (mathematical model) whose optimum (minimum or maximum) is sought during the optimisation process.
3. Parameter (usually used in the plural) – input values of a function that determine the value of the variable. Searching for an optimum consists of selecting parameter values that produce the optimum.
4. Constraints (or rules) – mathematically formulated rules that restrict the region in which the optimum is sought.
5. Index – because variables and parameters are often represented using the same symbol, they are usually additionally indexed numerically to make mathematical notation unambiguous.

## Examples

### 1. Financial investments

We have PLN 100,000 available and are looking for investment options that maximise profit. The following alternatives are available:

1. a low-risk investment with an annual return of 5%,
2. a medium-risk investment with an annual return of 10%,
3. a high-risk investment with an annual return of 15%.

Common sense suggests, however, that we should not allocate all of our capital to medium- or high-risk investments. Let us allocate at most 10% of the total capital to the high-risk investment and at most 20% to the medium-risk investment.

<ins>How can we formulate a mathematical model?</ins>

**Variables**, i.e. what we predict
- $R_1$ – profit from the first investment option
- $R_2$ – profit from the second investment option
- $R_3$ – profit from the third investment option

**Parameters**, i.e. how we control the prediction externally
- $C_1$ – amount allocated to the first investment option
- $C_2$ – amount allocated to the second investment option
- $C_3$ – amount allocated to the third investment option

**Constraints**, i.e. the rules followed by the model
- $C_1 + C_2 + C_3 = 100000$
- $R_1 = 0.05 C_1$
- $R_2 = 0.1 C_2$
- $R_3 = 0.15 C_3$
- $0 \leq C_2 \leq 0.2 \times 100000$
- $0 \leq C_3 \leq 0.1 \times 100000$

**Objective function**
- $max(R_1 + R_2 + R_3)$


The whole model can be written more compactly, avoiding unnecessary repetition, especially when the number of criteria is large.

**Index set**
- $F = \{1, 2, 3\}$

**Constraints**
- $\sum_{f \in F} C_f = 100000$
- $R_f = return(C_f) \forall f \in F$
- $C_f^{min} \leq C_f \leq C_f^{max} \forall f \in F$

**Objective function**
- $max \sum_{f \in F} R_f$

### 2. Table production costs

Consider a production process for kitchen tables, whose components are manufactured by three machines. Our task is to choose the number of parts produced by each machine so as to minimise the production cost, assuming that the target for the next month is to produce 1,000 tables.

The production cost of each machine can be expressed as follows:
1. $C_1 = 0.1P_1^2 + 0.5P_1 + 0.1$
2. $C_2 = 0.3P_1 + 0.5$
3. $C_3 = 0.01 P_3^3$

where $P_i$ denotes the number of products manufactured by the $i$-th machine.

**Variables**
- $C_1$ – production cost of machine 1
- $C_2$ – production cost of machine 2
- $C_3$ – production cost of machine 3

**Parameters**
- $P_1$ – number of products from machine 1
- $P_2$ – number of products from machine 2
- $P_3$ – number of products from machine 3

**Constraints**
- $P_1 + P_2 + P_3 = 1000$
- $C_1 = 0.1P_1^2 + 0.5P_1 + 0.1$
- $C_2 = 0.3P_2 + 0.5$
- $C_3 = 0.01 P_3^3$
- $P_1, P_2, P_3 \geq 0$

**Objective function**
- $min(C_1 + C_2 + C_3)$

How can this be expressed more compactly and in a more general form?

**Index set**
- $F = \{1, 2, 3\}$

**Constraints**
- $\sum_i P_i = 1000$
- $C_1 = 0.1 P_1^2 + 0.5 P_1 + \beta_1 0.1$
- $C_2 = 0.3P_2 + \beta_2 0.5$
- $C_3 = 0.01 P_3^3$
- $P_1 \leq \beta_1 M$
- $P_2 \leq \beta_2 M$
- $P_1, P_2, P_3 \geq 0$

**Objective function**
- $min \sum_i C_i$

Additional symbols
- $M$ – an arbitrarily large number representing the fixed cost of operating the machine
- $\beta_j$ – a binary variable representing whether the machine is switched on

It is worth noting that the operating cost of machines 1 and 2 depends not only on the number of parts produced. The constant term of the polynomial defining the cost is an additional fixed cost of operating the machine, regardless of the number of parts produced. Even when production is zero, this cost is incurred solely because the machine is switched on.

**Variables**
- $X_{i,j}$ – binary decision indicating adjacency (a direct connection) between nodes $i$ and $j$

**Parameters**
- $D_{i,j}$ – travel cost from node $i$ to node $j$

**Sets**
- $\Omega_i^{in}$ – set of incoming nodes for node $i$
- $\Omega_i^{out}$ – set of outgoing nodes for node $i$

**Objective function**
- $min \sum_{i,j} X_{i,j}D_{i,j}$

**Constraints**
- $\sum_{j \in \Omega_{A}^{out}} X_{A,j} = 1$
- $\sum_{i \in \Omega_{B}^{in}} X_{A,j} = 1$
- $\sum_{j \in \Omega_{i}^{out}} X_{i,j}= \sum_{j \in \Omega_{i}^{in}} X_{j,i} \forall_i \setminus \{A,B\}$
