# pandas library: data processing

[pandas](https://pandas.pydata.org) is a Python library for processing and visualising data. Thanks to its convenient and clear interface, it is one of the most widely used tools in data analysis.

## Installation

Install pandas with `pip install pandas`.

## Getting started with pandas

pandas is an external library, so the first step is to import it:

```python
import pandas as pd  
```

### Loading data

Data to be analysed must first be loaded from an external source. CSV is one of the most common formats for transferring structured data. CSV files can be loaded using the pandas `read_csv` function.

```python
movie: pd.DataFrame = pd.read_csv("data/movie.csv")
```

The data from the CSV file have been loaded into a `DataFrame` object, i.e. a tabular data frame. In a Jupyter notebook, the loaded data can be displayed directly:

```python
movie  
```

A data frame consists of named columns and row indices, here ranging from 0 to 4915.

When a data frame contains many rows, the `head` method can be used to display only the first $n$ rows:

```python
movie.head(n=3)  
```

Similarly, `tail` displays the last $n$ rows:

```python
movie.tail(n=4) 
```

The `movie.csv` file was loaded using `read_csv` with only a path specified. The `read_csv` function provides many additional options for customising how data are loaded. A CSV file is simply a text file in which values are separated by a delimiter. CSV files may also contain column headers, which are represented as a row in the file. The example below loads the data from `movie.csv` while explicitly specifying properties such as a comma separator and a header in row 0:

```python
movie: pd.DataFrame = pd.read_csv("data/movie.csv", sep=",", header=0)  
```

The `sep` argument specifies the value separator in the CSV file; its default is a comma. The `header` argument specifies the row containing column names. If no header exists, pass `None`; column names can then be assigned separately.

### Representing columns and rows in a data frame

Use the `columns` attribute of a `DataFrame` to obtain the list of column names:

```python
movie.columns  
```

Row indices (or row names) can be obtained in a similar way:

```python
movie.index 
```

In the movie data set, rows use numerical indices from 0 to 4915 inclusive. A data frame can, however, use arbitrary row indices, including text labels.

### Data types in data-frame columns

Use the `dtypes` attribute of a `DataFrame` to inspect the data types stored in its columns:

```python
movie.dtypes  
```

The `value_counts` method counts the frequency of individual values. It can, for example, be used to summarise the frequencies of data types across columns:

```python
movie.dtypes.value_counts()  
```

### Statistical summaries of data-frame columns

Calling `describe` on a `DataFrame` returns basic descriptive statistics for columns containing numerical values:

```python
movie.describe()  
```

## Selecting columns

There are several ways to select columns from a data frame:

```python
col: pd.Series = movie["genres"]  

col
```

```python
movie.genres 
```

Columns in a data frame are `Series` objects, meaning that their values are associated with the row indices of the data frame. Both column-selection styles shown above have the same effect; however, dot notation cannot be used for column names containing, for example, spaces.

Multiple columns can be selected by passing a list of their names:

```python
movie[["actor_1_name", "actor_2_name", "actor_3_name"]].head()  
```

### Selecting columns by data type

Columns can also be selected according to the kind of data they contain using the `select_dtypes` method of a `DataFrame`.

```python
movie.select_dtypes(include=["float"]).head()  
```

### Selecting columns by a phrase in the name

Calling `filter` on a `DataFrame` can return columns whose names contain a specified phrase using the `like` argument:

```python
movie.filter(like="facebook").head()  
```

## Working with missing values

Some values in data-frame columns may be missing. Calling `isnull` on a `Series` returns a Boolean series indicating whether each value in the column is missing:

```python
movie["genres"].isnull()  
```

Calling `any` on the resulting object allows us to check whether at least one value is true:

```python
movie["genres"].isnull().any()  
```

The `all` method checks whether all values are true.

If a column contains missing values, they can be filled with a specified value, for example zero:

```python
movie["actor_1_facebook_likes"].fillna(0)  
```

To replace missing values in the original data frame, call `fillna` on the `DataFrame` and pass a dictionary mapping column names to replacement values:

```python
movie.fillna({  
    "actor_1_facebook_likes": 200,
    "actor_3_facebook_likes": 100,
})  
```

## Arithmetic on data series

Standard arithmetic operators can be applied to `Series` objects. For example, the value 1 can be added to the first five rows of the `imdb_score` column:

```python
movie["imdb_score"].head()  
```

```python
movie["imdb_score"].head() + 3 
```

Other arithmetic operations, including subtraction, multiplication, and division, can be performed in the same way.

## Logical operations on data series

Logical operations on data-frame values produce Boolean results. For example, the following checks whether the first five values in `imdb_score` are greater than 7.5:

```python
movie["imdb_score"].head() > 7.5  
```

Other logical comparisons, such as greater-than and less-than comparisons, can be performed in the same way.

## Setting a data-frame index

The `set_index` method makes a selected column the index of the data frame:

```python
movie_new_idx: pd.DataFrame = movie.set_index("movie_title")  

movie_new_idx
```

The same result can be achieved when loading the CSV file with `read_csv` by passing the column name to `index_col`.

## Renaming columns and row indices

Use the `rename` method of a `DataFrame` and pass dictionaries mapping old names to new ones:

```python
from typing import Dict

idx_rename: Dict[str, str] = {  
    "Avatar": "Avatar II",
    "Spectre": "Spectre II",
}

col_rename: Dict[str, str] = {  
    "director_facebook_likes": "Director FB likes",
    "num_critic_for_reviews": "Num critic for reviews",
}

movie_new_idx = movie_new_idx.rename(index=idx_rename, columns=col_rename)  

movie_new_idx.head(n=7)  
```

## Creating new columns

A column containing a constant value, for example 100, can be added as follows. The column will appear at the end of the data frame:

```python
movie_new_idx["visits_num"] = 100  

movie_new_idx.head()
```

pandas also allows new columns to be created from existing ones. For example, the following column sums all likes received by actors and the director:

```python
movie_new_idx["total_facebook_likes"] = (movie_new_idx["actor_1_facebook_likes"] + movie_new_idx["actor_2_facebook_likes"] + movie_new_idx["actor_3_facebook_likes"] + movie_new_idx["Director FB likes"])

movie_new_idx.head() 
```

## Removing columns

The `del` statement can be used to remove columns from a data frame:

```python
del movie_new_idx["color"]  

movie_new_idx.head()
```

## Selecting rows

The `loc` indexer of a data frame selects one or more rows by index:

```python
movie_new_idx.loc["Avatar II"]
```

```python
movie_new_idx.loc[["Avatar II", "Spectre II"]]  
```

If a data frame has a numerical index, numerical row labels can simply be supplied to `loc`:

```python
movie.loc[[0, 1, 2]]  
```

The `iloc` indexer, similarly to NumPy indexing, selects rows and columns by their integer positions:

```python
movie_new_idx.iloc[0:10, :]  
```

```python
movie_new_idx.iloc[:, 3:7]  
```
