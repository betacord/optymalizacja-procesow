1. GLPK
   1. conda (zalecane): `conda install -c conda-forge glpk`
   2. pip (niezalecane - brak stabilności): `pip install pulp`
   3. alternatywnie: instalacja poprzez popbranie pliku wykonywalnego `https://www.gnu.org/software/glpk/` oraz wskazanie jego lokalizacji parametrem `executable` klasy `SolverFactory`

   3. alternatywnie: instalacja poprzez popbranie pliku wykonywalnego `https://coin-or.github.io/Ipopt/INSTALL.html` oraz wskazanie jego lokalizacji parametrem `executable` klasy `SolverFactory`
